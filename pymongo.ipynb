```python
import pandas as pd

file_path = "NorthStar_Full_Dataset.xlsx"

customers_df = pd.read_excel(file_path, sheet_name="customers")
orders_df = pd.read_excel(file_path, sheet_name="orders")
deliveries_df = pd.read_excel(file_path, sheet_name="deliveries")
drivers_df = pd.read_excel(file_path, sheet_name="drivers")
complaints_df = pd.read_excel(file_path, sheet_name="complaints")
```

```python
customers_df = customers_df.dropna()
orders_df = orders_df.dropna()
deliveries_df = deliveries_df.dropna()
drivers_df = drivers_df.dropna()
complaints_df = complaints_df.dropna()
```

```python
customers_data = customers_df.to_dict("records")
orders_data = orders_df.to_dict("records")
deliveries_data = deliveries_df.to_dict("records")
drivers_data = drivers_df.to_dict("records")
complaints_data = complaints_df.to_dict("records")
```

```python
db.deliveries.create_index("customer_id")
db.deliveries.create_index("driver_id")
db.deliveries.create_index("hub_id")
db.deliveries.create_index("delivery_status")
```

```python
db.customers.insert_many(customers_data)
db.orders.insert_many(orders_data)
db.deliveries.insert_many(deliveries_data)
db.drivers.insert_many(drivers_data)
db.complaints.insert_many(complaints_data)
```
